```python
# Creating an empty dictionary 
empty_dict = {}
print("Empty dictionary:", empty_dict)
print("Type of empty_dict:", type(empty_dict)) 
# Creating an empty dictionary using dict()
empty_dict2 = dict()
print("Empty dictionary using dict():", empty_dict2)
```

```python
# Creating a dictionary with initial values
student_info = {    "name": "John Doe",    "age": 20,    "grade": "A" }
print("Student Info:", student_info)
```

```python
# Creating a dictionary from a list of tuples
subjects = [("Math", 85), ("Science", 90), ("English", 78)]
subject_dict = dict(subjects)
print("Subject Dictionary:", subject_dict)
```

```python
# Creating a dictionary using dict() with keyword arguments 
person = dict(name="Alice", age=25, city="New York")
print("Person Dictionary:", person)
```

```python
# Using dict() to create a dictionary
grades = dict(Math=85, Science=90, English=78)
print("Grades Dictionary:", grades)
```

```python
# Finding the length of a dictionary 
student_info = {"name": "John Doe", "age": 20, "grade": "A"}
print("Length of student_info:", len(student_info))
```

```python
# Clearing a dictionary 
student_info = {"name": "John Doe", "age": 20, "grade": "A"}
student_info.clear()
print("Student Info after clear:", student_info)
```

```python
# Using get() to access dictionary values
student_info = {"name": "John Doe", "age": 20, "grade": "A"}
print("Name:", student_info.get("name"))
print("GPA:", student_info.get("GPA", "Not Available"))
```

```python
# Popping a value from the dictionary
student_info = {"name": "John Doe", "age": 20, "grade": "A"} 
age = student_info.pop("age")
print("Popped Age:", age)
print("Student Info after pop:", student_info)
```

```python
# Popping the last inserted item 
student_info = {"name": "John Doe", "age": 20, "grade": "A"} 
last_item = student_info.popitem()
print("Popped Item:", last_item)
print("Student Info after popitem:", student_info)
```

```python
# Getting all keys from the dictionary 
student_info = {"name": "John Doe", "age": 20, "grade": "A"}
keys = student_info.keys()
print("Keys:", keys)
```

```python
# Getting all values from the dictionary 
student_info = {"name": "John Doe", "age": 20, "grade": "A"} 
values = student_info.values()
print("Values:", values)
```

```python
# Getting all items from the dictionary 
student_info = {"name": "John Doe", "age": 20, "grade": "A"} 
items = student_info.items()
print("Items:", items)
```

```python
# Copying a dictionary 
student_info = {"name": "John Doe", "age": 20, "grade": "A"} 
student_copy = student_info.copy()
print("Original Dictionary:", student_info)
print("Copied Dictionary:", student_copy)
```

```python
# Updating a dictionary with another dictionary
student_info = {"name": "John Doe", "age": 20} 
extra_info = {"grade": "A", "GPA": 3.9} 
student_info.update(extra_info)
print("Updated Student Info:", student_info)
```
