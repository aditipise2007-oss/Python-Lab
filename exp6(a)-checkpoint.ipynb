```python
#creating a empty set
empty_set = set()
print("Empty set:", empty_set)
print("Type of empty_set:", type(empty_set))
```

```python
#Create a set with element
fruits = {"apple","banana","cherry"}
print("set of fruits:", fruits)
```

```python
#cerat a set with mixed data type
mixed_set = {1,"A",3.14,True}
print("mixed set:", mixed_set)
```

```python
#creating a set from  list
list_set = ([1,2,3,4])
print("set from list:", list_set)
#cerating a set for a string
string_set = set("hello")
print("set from string:", string_set)
#cerating a set for a tuple
tuple_set = set((10,20,30))
print("set from tuple:", tuple_set)
```

```python
#adding an element to set
fruits = {"apple","banana","cherry"}
fruits.add("orange")
print("set after adding 'orange':", fruits)
```

```python
#adding multiple element to a set
fruits = {"apple","banana","cherry"}
fruits.update("mango","grapes")
print("set after update:", fruits)
```

```python
#copied a set
original_set = {"apple", "banana", "cherry"}
copied_set = original_set.copy()
print("Copied set:", copied_set)
```

```python
# Removing an arbitrary element from a set
fruits = {"apple","banana","cherry"}
popped_element = fruits.pop()
print("popped_element:", popped_element)
print("set after pop:", fruits)
```

```python
#removing the specific element in the set
fruits = {"apple","banana","cherry"}
fruits.remove("banana")
print("set after removing 'banana':", fruits)
```

```python
# Discarding an element from a set
fruits = {"apple","banana","cherry"}
fruits.discard("banana")
print("set after discarding 'banana':", fruits)
# Discarding a non-existent element
fruits.discard("orange")
print("set after discarding 'orange':", fruits)
```

```python
#clearing a set
fruits = {"apple","banana","cherry"}
fruits.clear()
print("set after clear:", fruits)
```

```python
#union of two set
set1 = {"apple","banana","cherry"}
set2 = {"mango","grapes","banana"}
union_set = set1.union(set2)
print("union of set1 and set2:", union_set)
#using the | operator
union_set_operator = set1 | set2
print("union using | operator:", union_set_operator)
```

```python
#intersection of two sets
set1 = {"apple","banana","cherry"}
set2 = {"mango","grapes","banana"}
intersection_set = set1.intersection(set2)
print("intersection of set1 and set2:", intersection_set)
#using & operation 
intersection_set_operation = set1 & set2
print("intersection using & operation:", intersection_set_operation)
```

```python
#difference between two set
set1 = {"apple","banana","cherry"}
set2 = {"mango","grapes","banana"}
difference_set = set1.difference(set2)
print("difference of set1 and set2:", difference_set)
#using - operation
difference_set_operation = set1 - set2
print("difference using - operation:", difference_set_operation)
```
